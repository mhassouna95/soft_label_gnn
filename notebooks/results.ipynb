```python
import os
import pickle
import sys
from pathlib import Path
from typing import Optional, Tuple
import numpy as np
import pandas as pd
```

```python
np.random.seed(8888)
seeds = list(np.random.randint(0, 10000, 20))
#get greedy results
collect_scores = {}
collect_times = {}
path = Path("../seeds_results/01_trial_698/seeds_gnn_95_trial_698")
for seed in seeds:
    try: 
        with open(f'{path}/seed_res_gnn_{seed}.pkl', 'rb') as handle:
            collect_scores.update(pickle.load(handle))

        with open(f'{path}/surv_time_gnn_{seed}.pkl', 'rb') as handle:
            collect_times.update(pickle.load(handle))
    except:
        print(seed)
```

```python
df_scores_soft = pd.DataFrame(collect_scores).T
print(df_scores_soft.mean())
```
